```python
#creation des copies des datasets
df_cmd = cmd.copy()
df_msg = msg.copy()
#verification des variables : cmd
df_cmd.columns = df_cmd.columns.str.strip().str.lower()
print(df_cmd.columns.tolist()) 

#verification des variables : msg
df_msg.columns = df_msg.columns.str.strip().str.lower()
print(df_msg.columns.tolist()) 


#separation des donnees numeriques et categoriels
df_cmd_num = df_cmd.select_dtypes(include = "number")

df_cmd_catg = df_cmd.select_dtypes(exclude="number")
```

```python
#discrete ou continue?
for i in df_cmd_num:
    if isinstance(i,int):
        print(f"\n{i} : numerique discrete | justification : nombre entier")
    print(f"\n{i} : numerique continue | valeur decimal")
#il n'exist que les variables numeriques continues
```

```python
"""les anomalies : les doublons"""
for i in df_cmd:
    ligne_doublons = df_cmd[i].duplicated().sum()
    print(f"{i} : Doublons = {ligne_doublons}")
#suppression des ID's doublons 
df_cmd = df_cmd.drop_duplicates(subset='id_commande')
df_cmd.shape
```

```python
df_cmd_catg.value_counts()
```

```python
#conversion de date en type datetime
df_cmd['date_commande'] = pd.to_datetime(df_cmd['date_commande'], format="mixed")

#verification de type de data de commande
df_cmd['date_commande'].dtype

#creation du colonne mois
df_cmd['mois'] = df_cmd['date_commande'].dt.month

#verification des colonnes
df_cmd.columns.tolist()
```

```python
"""les anomalies : les valeurs manquantes"""
#les valeurs numeriques
print('la somme des valeurs manquantes pour chaque variable numerique\n')
for i in df_cmd_num.columns:
    manquantes = df_cmd[i].isnull().sum()
    print(f"{i} : manquantes = {manquantes}")

print('-'*50)
#les valeurs categoriels
print('la somme des valeurs manquantes pour chaque variable categoriel\n')
for i in df_cmd_catg.columns:
    manquantes = df_cmd[i].isnull().sum()
    print(f"{i} : manquantes = {manquantes}")
```

```python
var = df_cmd.groupby(['nb_articles','categorie'])['montant_dh'].median()
```

```python
#les valeurs manquantes en age_client
median_age = df_cmd_num['age_client'].median()
df_cmd_num['age_client'] = df_cmd_num['age_client'].fillna(median_age)
df_cmd_num['age_client'].isna().sum()
```

```python
#les valeurs manquantes en nb_articles
mode_nb_article = statistics.mode(df_cmd_num['nb_articles'])
df_cmd_num['nb_articles'] = df_cmd_num['nb_articles'].fillna(mode_nb_article)
df_cmd_num['nb_articles'].isna().sum()
```

```python
#les valeurs manquantes en montant_dh
# median_montant = df_cmd_num['montant_dh'].median()
# df_cmd_num['montant_dh'] = df_cmd_num['montant_dh'].fillna(median_montant) 
# df_cmd_num['montant_dh'].isna().sum()
var = df_cmd.groupby(['nb_articles','categorie'])['montant_dh'].transform('median')
df_cmd_num['montant_dh'] = df_cmd_num['montant_dh'].fillna(var) 
df_cmd_num['montant_dh'].isna().sum()
```

```python
median_delai = df_cmd_num['delai_livraison_jours'].median()
df_cmd_num['delai_livraison_jours'] = df_cmd_num['delai_livraison_jours'].fillna(median_delai)
df_cmd_num['delai_livraison_jours'].isna().sum()
```

```python
mode_sexe = statistics.mode(df_cmd_catg['sexe'])
df_cmd_catg['sexe'] = df_cmd_catg['sexe'].fillna(mode_sexe)
df_cmd_catg['sexe'].isna().sum()
```

```python
mode_canal = statistics.mode(df_cmd_catg['canal'])
df_cmd_catg['canal'] = df_cmd_catg['canal'].fillna(mode_canal)
df_cmd_catg['canal'].isna().sum()
```

```python
mode_satisfaction = statistics.mode(df_cmd_catg['satisfaction'])
df_cmd_catg['satisfaction'] = df_cmd_catg['satisfaction'].fillna(mode_satisfaction)
df_cmd_catg['satisfaction'].isna().sum()
```
